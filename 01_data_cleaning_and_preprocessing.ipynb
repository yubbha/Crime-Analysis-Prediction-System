## 5. Date & Time Processing

### 5.1 Location data (IMPORTANT for clustering)

```python
df = df.dropna(subset=["LAT", "LON"])
```

### 5.2 Victim age (replace invalid values)

```python
df = df[df["Vict Age"] > 0]
```

### 5.3 Categorical fields (fill with "Unknown")

```python
df["Weapon Desc"] = df["Weapon Desc"].fillna("Unknown")
df["Vict Sex"] = df["Vict Sex"].fillna("Unknown")
df["Premis Desc"] = df["Premis Desc"].fillna("Unknown")
```

## 6. Feature Selection for Tasks

### 6.1 Convert crime date to datetime

```python
df["DATE OCC"] = pd.to_datetime(df["DATE OCC"], errors="coerce")
```

### Remove invalid dates:

```python
df = df.dropna(subset=["DATE OCC"])
```

### 6.2 Extract useful time features (Used in trend analysis & Useful for classification & regression)

```python
df["year"] = df["DATE OCC"].dt.year
df["month"] = df["DATE OCC"].dt.month
df["day_of_week"] = df["DATE OCC"].dt.day_name()
```

## 7. Remove Duplicates (If Any)

```python
df.duplicated().sum()
```

## 8. Select CLEAN FEATURES (Your Responsibility)

```python
selected_columns = [
    "year", "month", "day_of_week",
    "AREA NAME",
    "Crm Cd Desc",
    "Premis Desc",
    "Weapon Desc",
    "Vict Age",
    "Vict Sex",
    "LAT", "LON"
]

df_clean = df[selected_columns]
```

## 9. Save the Cleaned Dataset

### Saving the Cleaned Dataset
The cleaned dataset is saved for reuse in clustering, association rule mining,
classification, regression, and dashboard development.

```python
df_clean.to_csv("crime_data_cleaned.csv", index=False)

print("Cleaned dataset has been successfully saved as 'crime_data_cleaned.csv'")
print(f"Number of records saved: {df_clean.shape[0]}")
print(f"Number of features saved: {df_clean.shape[1]}")
```
