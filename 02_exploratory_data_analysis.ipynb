## 4. Crime Trend Analysis Over Time

Understanding how crime incidents change over time is important for identifying long-term
trends and supporting forecasting and regression tasks.

```python
crime_per_year = df_clean["year"].value_counts().sort_index()

plt.figure()
plt.plot(crime_per_year.index, crime_per_year.values)
plt.title("Crime Incidents per Year")
plt.xlabel("Year")
plt.ylabel("Number of Crimes")

plt.savefig(r"C:\Users\hp\Desktop\Data Mining\Project\figure_1_crime_trend.png",
            dpi=300, bbox_inches="tight")

plt.show()
```

### Observation
The number of reported crime incidents varies across different years, indicating the presence
of temporal trends. This suggests that time-based features are relevant for further analysis and
predictive modeling.

## 5. Distribution of Crime Types

This section examines the most frequently reported crime types to understand dominant crime
categories and class distribution.

```python
# Top 10 Crime Types
top_crimes = df_clean["Crm Cd Desc"].value_counts().head(10)

plt.figure()
top_crimes.plot(kind="bar")
plt.title("Top 10 Most Common Crime Types")
plt.xlabel("Crime Type")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")

plt.savefig(r"C:\Users\hp\Desktop\Data Mining\Project\figure_2_top_crime_types.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation
A small number of crime categories account for a large proportion of reported incidents,
indicating class imbalance. This is an important consideration for classification models.

## 6. Crime Distribution by Area

Analyzing crime distribution across different areas helps identify regions with higher crime
activity and supports hotspot clustering.

```python
top_areas = df_clean["AREA NAME"].value_counts().head(10)

plt.figure()
top_areas.plot(kind="bar")
plt.title("Top 10 Areas by Crime Count")
plt.xlabel("Area")
plt.ylabel("Number of Crimes")
plt.xticks(rotation=45, ha="right")

plt.savefig(r"C:\Users\hp\Desktop\Data Mining\Project\figure_3_top_areas.png", dpi=300, bbox_inches="tight")

plt.show()
```

### Observation
Crime incidents are unevenly distributed across different areas, suggesting the presence of
high-risk zones. This supports the use of clustering techniques for hotspot identification.

## 7. Crime Distribution by Day of the Week (for my better understanding)

This analysis explores how crime incidents vary across different days of the week.

```python
day_counts = df_clean["day_of_week"].value_counts()

plt.figure()
day_counts.plot(kind="bar")
plt.title("Crime Distribution by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Number of Crimes")

plt.savefig(r"C:\Users\hp\Desktop\Data Mining\Project\figure_4_crime_distribution_by Day of the Weeek.png", dpi=300, bbox_inches="tight")

plt.show()
```
